#### (ii) Extraction of Time-Domain Features

For each of the six time series in every instance, the following seven features are extracted: minimum, maximum, mean, median, standard deviation, first quartile, and third quartile.

```python
# Six time series columns from each dataset
time_series_columns = [
    "avg_rss12",
    "var_rss12",
    "avg_rss13",
    "var_rss13",
    "avg_rss23",
    "var_rss23"
]

# Extracting the features 
def extract_features(file_path):
    
    # Reading the CSV file
    df = pd.read_csv(
        file_path,
        skiprows=4,
        usecols=range(7)
    )

    features = []

    # Going through each of the 6 time series
    for column in time_series_columns:
        series = df[column]

        # Adding the required features
        features.extend([
            series.min(),
            series.max(),
            series.mean(),
            series.median(),
            series.std(),
            series.quantile(0.25),
            series.quantile(0.75)
        ])

    return features
```

```python
sample_features = extract_features(sample_path)

print("Number of features:", len(sample_features))
print(sample_features)
```

```python
feature_names = []

for i in range(1, 7):
    feature_names.extend([
        f"min{i}",
        f"max{i}",
        f"mean{i}",
        f"median{i}",
        f"std{i}",
        f"1st_quart{i}",
        f"3rd_quart{i}"
    ])

print(feature_names)
print("Number of columns:", len(feature_names))
```

```python
# Combining all training and testing files
all_files = train_files + test_files

all_features = []

# Extracting the features
for file_path in all_files:
    features = extract_features(file_path)
    all_features.append(features)

# Creating a Df with all extracted features
feature_df = pd.DataFrame(
    all_features,
    columns=feature_names
)

feature_df.index = range(1, len(feature_df) + 1)
feature_df.index.name = "Instance"

feature_df
```

#### (iii) Standard Deviations and 90% Bootstrap Confidence Intervals

```python
feature_std = feature_df.std()

feature_std
```

```python
# calculating 90% bootstrap confidence interval
def bootstrap_std_ci(data, n_bootstrap=10000, confidence=0.90):
    
    # Remove missing values
    data = data.dropna()
    
    bootstrap_stds = []

    # Creating 10,000 bootstrap samples
    for _ in range(n_bootstrap):
        sample = np.random.choice(
            data,
            size=len(data),
            replace=True
        )

        # Calculate standard deviation of each sample
        bootstrap_stds.append(np.std(sample, ddof=1))

    # Find the lower and upper percentiles
    alpha = 1 - confidence

    lower = np.percentile(
        bootstrap_stds,
        100 * (alpha / 2)
    )

    upper = np.percentile(
        bootstrap_stds,
        100 * (1 - alpha / 2)
    )

    return lower, upper
```

```python

np.random.seed(42)

results = []

# Calculating the standard deviation and CI for all features
for feature in feature_df.columns:
    
    data = feature_df[feature]
    
    # Standard deviation
    std = data.std()
    
    # 90% bootstrap CI
    lower, upper = bootstrap_std_ci(data)
    
    results.append([
        std,
        lower,
        upper
    ])

#results
bootstrap_results = pd.DataFrame(
    results,
    index=feature_df.columns,
    columns=[
        "Standard Deviation",
        "90% CI Lower",
        "90% CI Upper"
    ]
)

bootstrap_results.round(4)
```

## 2. Linear Regression vs. Cubic Regression

### (a) Training RSS When the True Relationship is Linear

The cubic Regression is expected to have a lower or equal RSS because the linear model can fit inside the cubic model, so it can always perform at least as well as the linear model. Even though the true relationship is linear, X^2 and X^3 will pick up some random noise in these 100 points, which would push the training error down.

### (b) Test RSS When the True Relationship is Linear

The linear regression is expected to have a lower RSS because the true relationship is linear. The additional X^2 and X^3 in the cubic model has more flexibility and it would overfit into  random noise, which will cause overfitting. When the true relationship is linear, this overfitting increases the variance of the fitted model, which gives the cubic model a higher expected test error than the linear model.

### (c) Training RSS When the True Relationship is Nonlinear

The cubic regression is expected to have a training RSS less than or equal to that of the linear regression because the cubic model is more flexible, so it can fit the training observations at least as well as the linear model.

### (d) Test RSS When the True Relationship is Nonlinear

There is not enough information to determine which model will have the lower test RSS. 
If the true relationship is not linear(curve, cubic-like), the linear model has large bias, and the cubic model's lower bias outweighs its extra variance, so the cubic model has the lower test RSS.
If the true relationship is closer to linear than cubic, the bias removed by the cubic model is small compared with the variance it adds, so the linear model can still have the lower test RSS.
